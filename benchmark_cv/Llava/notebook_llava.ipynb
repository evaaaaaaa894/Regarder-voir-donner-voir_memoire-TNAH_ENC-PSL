```python
"""
LLaVA-1.5 : detection de mots-cles controles (thesaurus) + localisation via
Grounding DINO, sur une image.

Utilise l'implementation NATIVE de LLaVA dans transformers (classe
LlavaForConditionalGeneration), avec le checkpoint llava-hf/llava-1.5-7b-hf.
LLaVA-1.5 n'a pas de grounding natif : la localisation des termes valides est
donc deleguee a Grounding DINO (IDEA-Research/grounding-dino-tiny), un modele
de detection d'objets a vocabulaire ouvert, en entree duquel on passe les
termes confirmes par LLaVA.

Setup recommande (une seule fois, dans un notebook Colab tout neuf, GPU T4 ou mieux) :
    !pip install -q transformers==4.57.1 accelerate bitsandbytes torch torchvision pillow
Puis Runtime > Restart session avant d'executer ce script si vous changez de version.
"""

import json
import re
from PIL import Image, ImageDraw, ImageFont
import torch
from transformers import (
    AutoProcessor,
    LlavaForConditionalGeneration,
    BitsAndBytesConfig,
    AutoModelForZeroShotObjectDetection,
)

# ----------------------------------------------------------------------
# Config
# ----------------------------------------------------------------------
IMAGE_PATH = "Celestins-Qu-2b_4__68.jpg"
THESAURUS_PATH = "thesau_angl.json"
MODEL_ID = "llava-hf/llava-1.5-7b-hf"
GROUNDING_MODEL_ID = "IDEA-Research/grounding-dino-tiny"
MAX_NEW_TOKENS = 300
OUTPUT_IMAGE_PATH = "celestin_2b_llava_annotated.png"
OUTPUT_TXT_PATH = "celestin_2b_llava_results.txt"

TERM_FIELD = "Term_angl"  # thesaurus en anglais ; utilise "Term_TMS" pour la version francaise

MAX_WORDS_PER_BATCH = 150  # decoupe le thesaurus si trop long pour un seul prompt
BOX_THRESHOLD = 0.35
TEXT_THRESHOLD = 0.25

device = "cuda" if torch.cuda.is_available() else "cpu"

# ----------------------------------------------------------------------
# 1. Chargement du thesaurus (meme logique que le notebook Florence-2)
# ----------------------------------------------------------------------
with open(THESAURUS_PATH, "r", encoding="utf-8") as f:
    thesaurus_raw = json.load(f)

words = []
term_to_category = {}
if isinstance(thesaurus_raw, list):
    for fiche in thesaurus_raw:
        if isinstance(fiche, dict) and TERM_FIELD in fiche and isinstance(fiche[TERM_FIELD], str):
            term = fiche[TERM_FIELD].strip()
            if term:
                words.append(term)
                term_to_category[term.lower()] = fiche.get("cluster", "")
else:
    raise ValueError("Structure inattendue : une liste de fiches etait attendue.")

words = sorted(set(words))
print(f"{len(words)} termes charges depuis le thesaurus.")

# ----------------------------------------------------------------------
# 2. Chargement de LLaVA en 8-bit
# ----------------------------------------------------------------------
quant_cfg = BitsAndBytesConfig(load_in_8bit=True)

model = LlavaForConditionalGeneration.from_pretrained(
    MODEL_ID,
    quantization_config=quant_cfg,
    device_map="cuda:0",
)
processor = AutoProcessor.from_pretrained(MODEL_ID)

image = Image.open(IMAGE_PATH).convert("RGB")

# ----------------------------------------------------------------------
# 3. Interrogation de LLaVA (prompt instruction-following, comme Molmo2)
# ----------------------------------------------------------------------
def ask_llava(prompt_text):
    conversation = [
        {
            "role": "user",
            "content": [
                {"type": "image"},
                {"type": "text", "text": prompt_text},
            ],
        },
    ]
    text_prompt = processor.apply_chat_template(conversation, add_generation_prompt=True)
    inputs = processor(text=text_prompt, images=image, return_tensors="pt").to(model.device)

    with torch.inference_mode():
        output_ids = model.generate(
            **inputs,
            max_new_tokens=MAX_NEW_TOKENS,
            do_sample=False,
        )

    new_tokens = output_ids.shape[1] - inputs["input_ids"].shape[1]
    print(f"[debug] tokens generes : {new_tokens}")

    decoded = processor.batch_decode(output_ids, skip_special_tokens=True)[0]

    if "ASSISTANT:" in decoded:
        return decoded.split("ASSISTANT:")[-1].strip()
    return decoded.strip()


def build_batches(word_list, max_words_per_batch):
    return [word_list[i:i + max_words_per_batch]
            for i in range(0, len(word_list), max_words_per_batch)]


def build_prompt(word_batch):
    word_list_str = ", ".join(word_batch)
    return (
        "Here is a closed list of keywords: "
        f"{word_list_str}.\n"
        "Look carefully at the image and answer ONLY with the words from this list that "
        "correspond to elements actually visible in the image, separated by commas. "
        "Do not use any word outside this list, do not repeat words, do not add any explanation. "
        "If none of the words apply, answer: none."
    )


batches = build_batches(words, MAX_WORDS_PER_BATCH)

all_matches = set()
raw_answers = []

for i, batch in enumerate(batches):
    prompt = build_prompt(batch)
    answer = ask_llava(prompt)
    raw_answers.append(answer)

    # Filtrage anti-hallucination : ne garder que les mots reellement dans le thesaurus,
    # matching mot-entier / expression, insensible a la casse.
    matches = []
    for w in batch:
        pattern = r"\b" + re.escape(w) + r"\b"
        if re.search(pattern, answer, flags=re.IGNORECASE):
            matches.append(w)
    all_matches.update(matches)

    print(f"\nLot {i + 1}/{len(batches)} ({len(batch)} termes) :")
    print("Reponse brute :", answer)
    print(f"-> {len(matches)} mot(s) valide(s) : {matches}")

validated = sorted(all_matches)

print("\n=== Mots valides par LLaVA (presents dans le thesaurus) ===")
for term in validated:
    print(f"- {term} ({term_to_category.get(term.lower(), '')})")

# ----------------------------------------------------------------------
# 4. Liberation memoire avant de charger Grounding DINO
# ----------------------------------------------------------------------
import gc
del model
gc.collect()
torch.cuda.empty_cache()

# ----------------------------------------------------------------------
# 5. Localisation des termes valides avec Grounding DINO
# ----------------------------------------------------------------------
grounding_processor = AutoProcessor.from_pretrained(GROUNDING_MODEL_ID)
grounding_model = AutoModelForZeroShotObjectDetection.from_pretrained(GROUNDING_MODEL_ID).to(device)

# Grounding DINO attend du langage naturel : on remplace "a/b" par "a or b" pour eviter
# de perturber le tokenizer avec des mots composes inhabituels.
def clean_for_grounding(term):
    return term.replace("/", " or ").lower()

query_to_term = {clean_for_grounding(t): t for t in validated}
grounding_text = ". ".join(query_to_term.keys()) + "." if validated else ""
print(f"[debug] requete Grounding DINO : {grounding_text!r}")

term_boxes = {}

if grounding_text:
    gd_inputs = grounding_processor(images=image, text=grounding_text, return_tensors="pt").to(device)
    with torch.inference_mode():
        gd_outputs = grounding_model(**gd_inputs)

    import inspect
    sig = inspect.signature(grounding_processor.post_process_grounded_object_detection)
    pp_kwargs = {}
    if "box_threshold" in sig.parameters:
        pp_kwargs["box_threshold"] = BOX_THRESHOLD
        pp_kwargs["text_threshold"] = TEXT_THRESHOLD
    else:
        pp_kwargs["threshold"] = BOX_THRESHOLD
        if "text_threshold" in sig.parameters:
            pp_kwargs["text_threshold"] = TEXT_THRESHOLD

    results = grounding_processor.post_process_grounded_object_detection(
        gd_outputs,
        gd_inputs.input_ids,
        target_sizes=[image.size[::-1]],
        **pp_kwargs,
    )[0]

    print(f"[debug] boxes brutes detectees : {len(results['boxes'])}")
    print(f"[debug] labels bruts : {results.get('text_labels', results.get('labels', []))}")

    for box, label in zip(results["boxes"], results.get("text_labels", results.get("labels", []))):
        label_str = str(label).lower().strip()
        matched_term = query_to_term.get(label_str)
        if matched_term is None:
            for query, term_orig in query_to_term.items():
                if label_str in query or query in label_str:
                    matched_term = term_orig
                    break
        if matched_term is None:
            print(f"[debug] label non reconnu, ignore : {label_str!r}")
            continue
        box_coords = [round(v, 1) for v in box.tolist()]
        term_boxes.setdefault(matched_term, []).append(box_coords)
        print(f"[{matched_term}] bounding box : {box_coords}")

del grounding_model
gc.collect()
torch.cuda.empty_cache()

# ----------------------------------------------------------------------
# 6. Rendu de l'image annotee
# ----------------------------------------------------------------------
annotated = image.copy()
draw = ImageDraw.Draw(annotated)

palette = [
    "#FF3B30", "#34C759", "#007AFF", "#FF9500", "#AF52DE",
    "#FF2D55", "#5AC8FA", "#FFCC00", "#4CD964", "#5856D6",
]

try:
    font = ImageFont.truetype("DejaVuSans-Bold.ttf", 22)
except Exception:
    font = ImageFont.load_default()

for i, (term, boxes) in enumerate(term_boxes.items()):
    color = palette[i % len(palette)]
    for box in boxes:
        x0, y0, x1, y1 = box
        draw.rectangle([x0, y0, x1, y1], outline=color, width=3)
        draw.text((x0 + 4, max(y0 - 24, 0)), term, fill=color, font=font)

annotated.save(OUTPUT_IMAGE_PATH)
print(f"\nImage annotee sauvegardee : {OUTPUT_IMAGE_PATH}")

from IPython.display import display
display(annotated)

# ----------------------------------------------------------------------
# 7. Sauvegarde des resultats texte
# ----------------------------------------------------------------------
with open(OUTPUT_TXT_PATH, "w", encoding="utf-8") as f:
    f.write(f"Image : {IMAGE_PATH}\n")
    f.write(f"Modele detection : {MODEL_ID}\n")
    f.write(f"Modele localisation : {GROUNDING_MODEL_ID}\n")
    f.write(f"Thesaurus : {THESAURUS_PATH} ({len(words)} termes)\n\n")

    f.write("=== Reponses brutes de LLaVA par lot ===\n")
    for i, ans in enumerate(raw_answers):
        f.write(f"[Lot {i + 1}] {ans.strip()}\n\n")

    f.write("=== Mots valides (presents dans le thesaurus) ===\n")
    f.write(", ".join(validated) + "\n\n")

    f.write("=== Bounding boxes par terme (pixels, x0 y0 x1 y1) ===\n")
    for term in validated:
        boxes = term_boxes.get(term, [])
        boxes_str = "; ".join(f"({b[0]}, {b[1]}, {b[2]}, {b[3]})" for b in boxes)
        f.write(f"{term}: {boxes_str if boxes_str else '(aucune bbox)'}\n")

print(f"\nResultats texte sauvegardes : {OUTPUT_TXT_PATH}")
```
